# Checkpoint 1 EDA

Quick validation of the cleaned and feature-engineered datasets. Adjust paths if you rerun the pipeline with custom output dirs.

```python
from pathlib import Path
import pandas as pd

processed_dir = Path('data/processed')
clean_path = processed_dir / 'cleaned_data.csv'
feat_path = processed_dir / 'feature_engineered_data.csv'

clean_df = pd.read_csv(clean_path, parse_dates=['match_date'])
feat_df = pd.read_csv(feat_path, parse_dates=['match_date'])

clean_df.head()
```

```python
# Basic missingness overview
missing = (clean_df.isna().mean() * 100).sort_values(ascending=False)
missing.head(20)
```

```python
# Correlation preview for engineered numerical columns
numeric_cols = [c for c in feat_df.columns if feat_df[c].dtype != 'object']
cor = feat_df[numeric_cols].corr(numeric_only=True)
cor[['y']].sort_values('y', ascending=False).head(20)
```

```python
# Sanity-check the GRU sequence shapes
import numpy as np
seq = np.load(processed_dir / 'gru_sequences.npz')
seq['X'].shape, seq['mask'].shape, seq['feature_names'][:5]
```
